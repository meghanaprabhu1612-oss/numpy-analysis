```python
"""

Student Marks Analysis using NumPy
Create an array containing marks for 10 students across 4 subjects. Your program should:
• Calculate each student's total marks.
• Calculate each student's average.
• Find the student index with the highest average.
• Find subject-wise averages using axis=0.
• Find students whose average is at least 60.
• Assign Pass/Fail based on an average threshold.
• Calculate the standard deviation of each subject.
• Normalize the subject scores using z-score standardization.
"""
```

```python
import numpy as np
arr=np.array([[85,90,78,92],[88,76,95,89],[90,85,80,87],[70,75,65,80],[95,92,88,94],[60,70,75,65],[82,88,90,85],[78,80,82,79],[92,95,90,93],[85,87,88,86]])
Total=np.sum(arr,axis=1) # Total marks for each student
Average=np.mean(arr,axis=1) # Average marks for each student
average2=np.max(np.mean(arr,axis=1)) # Highest average marks
Subject_averages=np.mean(arr,axis=0) # Subject-wise averages
Students_above_threshold=np.where(np.mean(arr,axis=1)>=60) # Students with average at least 60
Pass_Fail=np.where(np.mean(arr,axis=1)>=60,'Pass','Fail') # Assign Pass/Fail based on average threshold
std_dev=np.std(arr,axis=0) # Standard deviation of each subject
z = (arr - np.mean(arr,axis=0)) / np.std(arr,axis=0) 
print("Total Marks for each student:", Total)
print("Average Marks for each student:", Average)
print("Highest Average Marks:", average2)
print("Subject-wise Averages:", Subject_averages)
print("Students with Average at least 60:", Students_above_threshold)
print("Pass/Fail Status:", Pass_Fail)
print("Standard Deviation of Each Subject:", std_dev)
print("Z-Scores for Each Subject:", z)
# Normalize subject scores using z-score standardization
```
